## Model settings, toknizer

```python
checkpoint = "microsoft/git-base"
MAX_LEN = 3000
```

```python
from transformers import AutoProcessor, AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained(checkpoint)
tokenizer.model_max_length = MAX_LEN
```

```python
icons = "clean_icons"
svgs = []

for f in glob.glob(os.path.join("data", icons,"*.svg")):
  img = (f.split("/")[-1].split(".")[0])
  p = (os.path.join("data", icons, img+".svg"))
  svgs.append(open(p).read())
```

```python
# tokenize before
tokenizer.tokenize(svgs[0])
```

```python
def get_training_corpus():
  dataset = svgs
  for start_idx in range(0, len(dataset), 1000):
    samples = dataset[start_idx : start_idx + 1000]
    yield samples
```

```python
tokenizer.train_new_from_iterator(get_training_corpus(), 10000)
```

```python
# tokenize before
tokenizer.tokenize(svgs[1])
```

```python
nums = "1234567890"
for i in nums:
  print(i in tokenizer.vocab)

# all cool. we have all the numbers now
```
